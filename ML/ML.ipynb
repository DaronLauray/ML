```python
#model accuracy was at 100% but that is cause it was training on 
#future data and testing the past, so adding a buffer to fix that
#now time to clean data and set manual binary encoding for home team wins
games= pd.read_csv('./Data/Games.csv')
games= games.sort_index(ascending=False)
games_data= games[['hometeamId','awayteamId','winner']]
teams_names_key= games[['hometeamName','hometeamId']]
games.describe()
X= games_data.loc[:, games_data.columns !='winner']
y= games['winner']
#make team winner a binary value for training
winner_id=[]
for id in y:
    winner_id.append(id)

home_team_id=[]
for ID in games['hometeamId']:
    home_team_id.append(ID)

i=0
home_team_won=[]
while i<= len(winner_id)-1:
    if winner_id[i]==home_team_id[i]:
        home_team_won.append(int(1))
    else:
        home_team_won.append(int(0))
    i +=1


games['home_team_won']= home_team_won

#Making away team won column
away_team_won= []
for win in home_team_won:
    if win == 1:
       away_team_won.append(0)
    else:
       away_team_won.append(1)


games['away_team_won']= away_team_won
y=games[['home_team_won']].copy()

#model to test/train with
model= LogisticRegression()

#standard "train/test split" out of the box
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=.2, random_state=39)

#manual "train/test split" to account for future values in df
from sklearn.model_selection import TimeSeriesSplit
tscv= TimeSeriesSplit( n_splits=4, gap=0,max_train_size=14639,test_size=14639)

#test to make sure split is even; how I want it
#(uncomment below to initialize)
#for i, (train_index,test_index) in enumerate(tscv.split(X,y)):
    #print(f'Fold {i}:')
    #print(f'Train Index:{train_index}')
    #print(f'Test Index: {test_index}')


all_splits= list(tscv.split(X,y))
train_0,test_0= all_splits[0]
train_1,test_1= all_splits[1]
train_2,test_2= all_splits[2]
# train_3,test_3= all_splits[3]
# train_4,test_4= all_splits[4]

# fit the model algo
model.fit(X.iloc[train_2],y.iloc[train_2])

#model accuracy check

predictions= model.predict(X.iloc[test_2])
score= accuracy_score(y.iloc[test_2],predictions)

#notes:
# score is confirmed at 100% going to play with values just in case to make sure, 
#it may be due to team scores being known
#update: after adjustment score is at 80% which is more realistic
#Yup, because team scores were known it gave higher accuracy, now it is down to 47%
#increased training to 90, and test to 10, got 70% accuracy just based on team match up
#going to see if I can provide a synthetic input to predict a real game now
#real life predictor seems to always favor a team if they are home regardless of other team. 
#going to include away stats too to see if that helps
#yup there was data leakage, now it is down to 60% accuracy going to continue R&D
#Daron Lauray 


score
```
